## Data Structures Part 1: Lists, Tuples


```python
#list
my_list = [1, 2, 3, 4, 5]
```

```python
# append()
my_list.append(6)
print(my_list)
```

```python
# extend()
my_list.extend([7, 8, 9])
print(my_list)
```

```python
# insert()
my_list.insert(2, 10)
print(my_list)
```

```python
# remove()
my_list.remove(3)
print(my_list)
```

```python
# pop()
popped_element = my_list.pop(3)
print(popped_element)
print(my_list)
```

```python
# index()
index = my_list.index(5)
print(index)
```

```python
# count()
count = my_list.count(10)
print(count)
```

```python
# sort()
my_list.sort()
print(my_list)
```

```python
# reverse()
my_list.reverse()
print(my_list)
```

```python
# sorted()
my_list1 = [2,4,5,2,1,3,6,7,9,8,10]
my_list_ = sorted(my_list1)  # list will be sorted in ascending order
print(my_list_)  # expected output is: [1, 2, 5, 6, 7, 8, 9, 10]
```

```python
# Creating a tuple
my_tuple = (1, 2, 3, 4, 5)
```

```python
# count()
count = my_tuple.count(3)
print(count)
```

```python
# index()
index = my_tuple.index(4)
print(index)
```

```python
print(my_tuple[1])
```

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
```

```python
#add()
my_set.add(6)
print(my_set)
```

```python
#update
my_set.update({7,8,9})
print(my_set)
```

```python
#remove()
my_set.remove(3)
print(my_set)
```

```python
#discard()
my_set.discard(5)
print(my_set)
```

```python
#pop
popped_element = my_set.pop()
print(popped_element)
```

```python
#clear()
my_set.clear()
print(my_set)
```

```python
# Creating sets for methods demonstration
set1 = {1, 2, 3}
set2 = {3, 4, 5}
```

```python
#union()
union_set = set1.union(set2)
print(union_set)
```

```python
# union operator
union_set1 = set1 | set2
print(union_set1)
```

```python
#intersection()
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
#intersection operator
intersection_set1 = set1 & set2
print(intersection_set1)
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
```

```python
#difference
difference_set = set1.difference(set2)
print(difference_set)
```

```python
#difference
difference_set = set2.difference(set1)
print(difference_set)
```

```python
#symmetric difference
symmetric_difference = set1.symmetric_difference(set2)
print(symmetric_difference)
```
